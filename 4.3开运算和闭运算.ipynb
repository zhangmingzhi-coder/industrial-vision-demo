```python
import cv2
import numpy as np

img = cv2.imread('ceshi4.png', 0)
kernel = np.ones((5, 5), np.uint8)

#开运算：先腐蚀后膨胀
opening = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)

#先膨胀后腐蚀
closing = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)

#显示对比
cv2.imread('Original', img)
cv2.imread('Opening', opening)
cv2.imread('Closing', closing)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('ceshi4.png', 0)
kernel = np.ones((5, 5), np.uint8)
img = cv2.bitwise_not(img)
#开运算：先腐蚀后膨胀
opening = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)

#先膨胀后腐蚀
closing = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)

#显示对比
fig, axes = plt.subplots(1, 3, figsize=(15,10))
axes[0].imshow(img)
axes[0].set_title("Original")
axes[0].axis('off')

axes[1].imshow(opening)
axes[1].set_title("Opening")
axes[1].axis('off')

axes[2].imshow(closing)
axes[2].set_title("Closing")
axes[2].axis('off')

plt.tight_layout()
plt.show()
# cv2.imread('Original', img)
# cv2.imread('Opening', opening)
# cv2.imread('Closing', closing)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
# 构造一张带有椒盐噪声的测试图
img = np.ones((200, 200), dtype=np.uint8) * 128
# 随机撒入黑白噪点
img[50, 50] = 0 
img[100, 100] = 0
img[51, 51] = 255
img[150, 150] = 255
# 高斯滤波处理椒盐噪声
gaussian_res = cv2.GaussianBlur(img, (5, 5), 0)
# 中值滤波处理椒盐噪声
median_res = cv2.medianBlur(img, 5)
cv2.imshow('Original Noise', img)
cv2.imshow('Gaussian on Noise', gaussian_res)
cv2.imshow('Median on Noise', median_res)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

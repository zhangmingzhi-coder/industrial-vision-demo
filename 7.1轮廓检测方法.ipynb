```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
#1.读取图像并转换为灰度图
img = cv2.imread('ceshi2.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

#2.图像二值化（轮廓检测的必要前置条件）
#阈值调整为127，最大值为255，采用THRESH_BINARY模式
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

#3.查找轮廓
#RETR_TREE获取所有轮廓以及完整层级关系
#CHAIN_APPROX_SAMPLE压缩水平、垂直和对角线方向的元素，只保留端点
contours, hierarchy = cv2.findContours(binary, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

#4.绘制轮廓（在原图上面绘制,颜色为绿色，线宽为2）
cv2.drawContours(img, contours, -1, (0, 0, 255), 2)

#显示结果
plt.imshow(img)
plt.axis('off')

plt.tight_layout()
plt.show()
```

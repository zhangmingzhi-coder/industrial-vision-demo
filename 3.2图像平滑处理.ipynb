```python
import cv2

img = cv2.imread('ceishi3.jpg')

#1.均值滤波：简单粗暴的灵芝平均
blur = cv2.blur(img, (5, 5))

#2.高斯滤波：中心权重高，周边权重低
gaussian = cv2.GaussianBlur(img, (5, 5), 0)

#3.中值滤波：取领域中间值，去除噪点的神器
median = cv2.medianBlur(img, 5)

#4.双边滤波：保边去噪，磨皮效果最好
bilateral = cv2.bilateralFilter(img, 9, 75, 75)

cv2.imshow('Original', img)
cv2.imshow('Blur', blur)
cv2.imshow('Gaussian', gaussian)
cv2.imshow('Median', median)
cv2.imshow('Bilateral', bilateral)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import matplotlib.pyplot as plt
img = cv2.imread('ceishi3.jpg')

#1.均值滤波：简单粗暴的灵芝平均
blur = cv2.blur(img, (5, 5))

#2.高斯滤波：中心权重高，周边权重低
gaussian = cv2.GaussianBlur(img, (5, 5), 0)

#3.中值滤波：取领域中间值，去除噪点的神器
median = cv2.medianBlur(img, 5)

#4.双边滤波：保边去噪，磨皮效果最好
bilateral = cv2.bilateralFilter(img, 9, 75, 75)

#创建一个2行3列的子图，并且设置画布大小
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

axes[0, 0].imshow(img)
axes[0, 0].set_title('Origianal')
axes[0, 0].axis('off')

axes[0, 1].imshow(blur)
axes[0, 1].set_title('Blur')
axes[0, 1].axis('off')

axes[0, 2].imshow(gaussian)
axes[0, 2].set_title('Gaussian')
axes[0, 2].axis('off')

axes[1, 0].imshow(median)
axes[1, 0].set_title('Median')
axes[1, 0].axis('off')

axes[1, 1].imshow(bilateral)
axes[1, 1].set_title('Bilateral')
axes[1, 1].axis('off')

axes[1, 2].axis('off')

#统一显示所有图片
plt.tight_layout()
plt.show()
# cv2.imshow('Original', img)
# cv2.imshow('Blur', blur)
# cv2.imshow('Gaussian', gaussian)
# cv2.imshow('Median', median)
# cv2.imshow('Bilateral', bilateral)
# cv2.waitKey(0)
# cv2.destroyAllWindow()
```

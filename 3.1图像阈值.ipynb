```python
import cv2

img = cv2.imread('ceishi1.jpg', cv2.IMREAD_GRAYSCALE)
#gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  #(传入为灰度图)
#全局阈值：固定阈值为127
ret1, thresh1 = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

#Otsu's二值化：自动寻找最优阈值
ret2, thresh2 = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

#自适应阈值：根据局部区域计算阈值 
thresh3 = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

cv2.imshow('Original', img)
cv2.imshow('Global', thresh1)
cv2.imshow('Otsu', thresh2)
cv2.imshow('Adaptive', thresh3)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import matplotlib.pyplot as plt
im = cv2.imread('ceishi1.jpg')
img = cv2.imread('ceishi1.jpg', cv2.IMREAD_GRAYSCALE)

#全局阈值：固定阈值为127
ret1, thresh1 = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

#Otsu's二值化：自动寻找最优阈值
ret2, thresh2 = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

#自适应阈值：根据局部区域计算阈值 (传入为灰度图)
thresh3 = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

#使用matplotlib在网页中显示图片,设置画布的大小为12×10（宽×高）
fig, axs = plt.subplots(1, 5, figsize=(15, 5))

#显示原图
axs[0].imshow(im)
axs[0].set_title('Original')
axs[0].axis('off')

axs[1].imshow(thresh1)
axs[1].set_title('Glogal')
axs[1].axis('off')

axs[2].imshow(thresh2)
axs[2].set_title('Otsu')
axs[2].axis('off')

axs[3].imshow(thresh3)
axs[3].set_title('Adaptive')
axs[3].axis('off')

axs[4].imshow(img)
axs[4].set_title('huidutu')
axs[4].axis('off')

#统一显示所有图片
plt.tight_layout()  #自动调整字间距，防止标题重叠
plt.show()
# cv2.imshow('Original', img)
# cv2.imshow('Global', thresh1)
# cv2.imshow('Otsu', thresh2)
# cv2.imshow('Adaptive', thresh3)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

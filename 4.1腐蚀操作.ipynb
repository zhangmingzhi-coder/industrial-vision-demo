```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

#读取图像，0表示用灰度图的方式表示
img = cv2.imread('ceshi4.png', 0)

#定义一个5×5全1 的矩阵作为结构元素（卷积核）
kernel = np.ones((5,5), np.uint8)

#执行腐蚀操作 iterations = 2, 表示连续腐蚀2次
eroded = cv2.erode(img, kernel, iterations=5)


fig, axes = plt.subplots(1, 2, figsize=(15, 10))

axes[0].imshow(img)
axes[0].set_title('Origianal')
axes[0].axis('off')

axes[1].imshow(eroded)
axes[1].set_title('eroded')
axes[1].axis('off')

#统一显示所有图片
plt.tight_layout()
plt.show()
#显示原图和腐蚀后的图像
# cv2.imshow('original', img)
# cv2.imshow('Eroded', eroded) 
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

**造成这个原因是：腐蚀的原理是当5×5的方块在图像中滑动时，只要在5×5这个范围内有一点像素值很低的，就会把像素拉的越低**


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

#读取图像，0表示用灰度图的方式表示
img = cv2.imread('ceshi4.png', 0)

#经行颜色反转
img = cv2.bitwise_not(img)

#定义一个5×5全1 的矩阵作为结构元素（卷积核）
kernel = np.ones((5,5), np.uint8)

#执行腐蚀操作 iterations = 2, 表示连续腐蚀2次
eroded = cv2.erode(img, kernel, iterations=1)


fig, axes = plt.subplots(1, 2, figsize=(15, 10))

axes[0].imshow(img)
axes[0].set_title('Origianal')
axes[0].axis('off')

axes[1].imshow(eroded)
axes[1].set_title('eroded')
axes[1].axis('off')

#统一显示所有图片
plt.tight_layout()
plt.show()
```

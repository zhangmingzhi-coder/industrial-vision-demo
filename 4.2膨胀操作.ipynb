```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

#读取灰度图
img = cv2.imread('ceshi4.png', 0)

#定义5×5结构
kernel = np.ones((5, 5), np.uint8)

#进行反转
img = cv2.bitwise_not(img)

#执行膨胀操作
dilated = cv2.dilate(img, kernel, iterations=2)

fig, axes = plt.subplots(1, 2,figsize=(15,10))
axes[0].imshow(img)
axes[0].set_title('Original')
axes[0].axis('off')

axes[1].imshow(dilated)
axes[1].set_title('Dilated')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2

img = cv2.imread('people and dog.jpg')
if img is None:
    print("错误：找不到图片")
else:
    #通过数组切片获取ROI：img[起始y：结束y， 起始x：结束x]
    #x对应宽（列）， y对应高（行）
    rio = img[270:750, 500:1000]
    
    #显示ROI区域
    cv2.imshow('RIO Window', rio)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2

img = cv2.imread('people and dog.jpg')
if img is None:
    print("错误：找不到图片")
else:
    #通过数组切片获取ROI：img[起始y：结束y， 起始x：结束x]
    #x对应宽（列）， y对应高（行）
    rio = img[270:750, 500:1000].copy()  #独立副本
    
    #显示ROI区域
    cv2.imshow('RIO Window', rio)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

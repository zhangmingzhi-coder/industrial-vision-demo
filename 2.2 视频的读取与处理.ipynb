```python
import cv2

#读取视频文件，0表示读取默认摄像头
cap = cv2.VideoCapture('屏幕录制 2025-12-14 163712.mp4')

#循环读取每一帧
while cap.isOpened():
    ret, fram = cap.read()  #ret:是否读取成功  fram：当前帧图像

    if not ret:
        break;  #视频播放完毕或者读取失败，退出循环

    #对每一帧进行处理(转换为灰度图)
    gray_fram = cv2.cvtColor(fram, cv2.COLOR_BGR2GRAY)

    #显示处理后的帧
    cv2.imshow('fram', gray_fram)

    #按‘q’键退出
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break
    
    #释放资源
    cap.release()
    cv2.destroyAllWindows()
```

```python
import cv2
import matplotlib.pyplot as plt
from IPython.display import clear_output  #用来清楚上一帧，实现动态播放效果
import time

#读取视频文件，0表示读取默认摄像头
cap = cv2.VideoCapture('屏幕录制 2025-12-14 163712.mp4')

#检查视频是否能打开
if not cap.isOpened():
    print("视频无法加载")
else:
    #循环读取每一帧
    while cap.isOpened():
        ret, fram = cap.read()  #ret:是否读取成功  fram：当前帧图像
    
        if not ret:
            print("视频播放完毕")
            break  #视频播放完毕或者读取失败，退出循环
    
        #对每一帧进行处理,颜色空间转换(转换为灰度图)
        gray_fram = cv2.cvtColor(fram, cv2.COLOR_BGR2GRAY)

        #显示后的图片
        plt.figure(figsize=(12, 8))  #设置显示大小
        #显示处理后的帧
        plt.imshow(gray_fram, cmap='gray')
        plt.axis('off')
        clear_output(wait=True)  #清楚上一帧图片，实现“播放”效果
        plt.show()
        time.sleep(0.005)
        # # #按‘q’键退出
        # if cv2.waitKey(25) & 0xFF == ord('q'):
        #     break
        
#释放资源
cap.release()
# cv2.destroyAllWindows()
```

```python
import cv2
import matplotlib.pyplot as plt
from IPython.display import clear_output  #用来清楚上一帧，实现动态播放效果
import time

#读取视频文件，0表示读取默认摄像头
cap = cv2.VideoCapture('屏幕录制 2025-12-14 163712.mp4')

#检查视频是否能打开
if not cap.isOpened():
    print("视频无法加载")
else:
    #循环读取每一帧
    while cap.isOpened():
        ret, fram = cap.read()  #ret:是否读取成功  fram：当前帧图像
    
        if not ret:
            print("视频播放完毕")
            break  #视频播放完毕或者读取失败，退出循环
    
        #对每一帧进行处理,颜色空间转换(转换为灰度图)
        gray_fram = cv2.cvtColor(fram, cv2.COLOR_BGR2RGB)

        #显示后的图片
        plt.figure(figsize=(12, 8))  #设置显示大小
        #显示处理后的帧
        plt.imshow(gray_fram)
        plt.axis('off')
        clear_output(wait=True)  #清楚上一帧图片，实现“播放”效果
        plt.show()
        time.sleep(0.005)
        # # #按‘q’键退出
        # if cv2.waitKey(25) & 0xFF == ord('q'):
        #     break
        
#释放资源
cap.release()
# cv2.destroyAllWindows()
```

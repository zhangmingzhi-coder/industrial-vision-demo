```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('canny.jpg', 0)
# img = cv2.bitwise_not(img)
#高斯滤波降噪
blurred = cv2.GaussianBlur(img, (5, 5), 1.4)

#canny边缘检测（参数：图像、低阈值、高阈值）
edges1 = cv2.Canny(blurred, 50, 150)
edges2 = cv2.Canny(blurred, 30, 100)


# v2.imshow('Original', img)
# cv2.imshow('blurred', blurred)
# cv2.imshow('edges', edges1)

#1.将两张边缘图像横向拼接
res = np.hstack((edges1, edges2))

#2.获取原图像的尺寸（高度， 宽度）
h, w = res.shape[:2]

#3.设定缩放比例
scale = 0.2
new_w = int(w*scale)
new_h = int(h*scale)
dsize = (new_w, new_h)

#4.缩放图片
res_resized = cv2.resize(res, dsize)

#5，正常创建窗口并显示
cv2.imshow('res', res_resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

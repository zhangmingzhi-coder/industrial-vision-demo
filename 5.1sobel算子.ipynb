```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('soble.jpg', 0)

#计算X方向的梯度（检查垂直边缘），输出深度设为-1表示与原图相同
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)

#计算y方向的梯度（检测水平边缘）
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

#计算梯度幅值
sobelxy = cv2.magnitude(sobelx, sobely)

sobelx = np.uint8(np.absolute(sobelx))
sobely = np.uint8(np.absolute(sobely))
sobelxy = np.uint8(sobelxy)

fig, axes = plt.subplots(2, 2, figsize=(15,10))
axes[0][0].imshow(img)
axes[0][0].set_title('Original')
axes[0][0].axis('off')

axes[0][1].imshow(sobelx)
axes[0][1].set_title('Sobelx')
axes[0][1].axis('off')

axes[1][0].imshow(sobely)
axes[1][0].set_title('Sobely')
axes[1][0].axis('off')

axes[1][1].imshow(sobelxy)
axes[1][1].set_title('Sobelxy')
axes[1][1].axis('off')

plt.tight_layout()
plt.show()
```

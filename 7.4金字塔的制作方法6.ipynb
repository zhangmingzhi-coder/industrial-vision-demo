```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('ceishi3.jpg')

#向下取样分辨率下降
down1 = cv2.pyrDown(img)
down2 = cv2.pyrDown(down1)
down3 = cv2.pyrDown(down2)
down4 = cv2.pyrDown(down3)

#向上取样分辨率上升512——>1024
up1 = cv2.pyrUp(img)
up2 = cv2.pyrUp(up1)


fig, axes = plt.subplots(1, 3, figsize=(15, 10))

axes[0].imshow(down4)
axes[0].set_title('Down Sample')
axes[0].axis('off')

axes[1].imshow(up2)
axes[1].set_title('Up Sample')
axes[1].axis('off')

axes[2].imshow(img)
axes[2].set_title('Sample')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
from IPython.display import Image, display

img = cv2.imread('people and dog.jpg')

#限制宽度为300像素，高度自动适应
display(Image(filename='people and dog.jpg',width=300))
#查看图像形状：（高，宽，通道）
print("图像形状:",img.shape)

#获取图像像素总数
print("像素总数:",img.size)

#获取图像数据类型
print("数据类型:",img.dtype)
    
cv2.imshow('people and dog',img)
cv2.waitKey(0)  #等待用户按键，0表示无限等待
cv2.destroyAllWindows()  #销毁所有窗口
```

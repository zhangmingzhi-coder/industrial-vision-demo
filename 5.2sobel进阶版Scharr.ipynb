```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('soble.jpg', 0)

#Scharr算子
Scharrx = cv2.Scharr(img, cv2.CV_64F, 1, 0)
Scharry = cv2.Scharr(img, cv2.CV_64F, 0, 1)
scharrxy = cv2.magnitude(Scharrx, Scharry)

#拉普拉斯算子
laplacian = cv2.Laplacian(img, cv2.CV_64F)

scharrx = np.uint8(np.absolute(Scharrx))
scharry = np.uint8(np.absolute(Scharry))
laplacian = np.uint8(np.absolute(laplacian))


fig, axes = plt.subplots(2, 2, figsize=(15,10))
axes[0][0].imshow(Scharrx)
axes[0][0].set_title('Scharrx')
axes[0][0].axis('off')

axes[0][1].imshow(Scharry)
axes[0][1].set_title('Scharry')
axes[0][1].axis('off')

axes[1][0].imshow(laplacian)
axes[1][0].set_title('Laplacian')
axes[1][0].axis('off')

axes[1][1].imshow(scharrxy)
axes[1][1].set_title('scharrxy')
axes[1][1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np

img = cv2.imread('ceishi1.jpg', 0)
kernel = np.ones((3, 3), np.uint8)

#计算形态学梯度
gradient = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, kernel)


cv2.imshow('Oeiginal', img)
cv2.imshow('Gradient', gradient)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('soble.jpg')
kernel = np.ones((3,3), np.uint8)
img = cv2.bitwise_not(img)
#计算形态学梯度
gradient = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, kernel)

fig, axes = plt.subplots(1, 2, figsize = (15, 10))
axes[0].imshow(img)
axes[0].set_title('Original')
axes[0].axis('off')

axes[1].imshow(gradient)
axes[1].set_title('Gradient')
axes[1].axis('off')

plt.tight_layout()
plt.show()
# cv2.imshow('Oeiginal', img)
# cv2.imshow('Gradient', gradient)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

```python
import cv2

img = cv2.imread('graduate.png')

#常熟填充：边缘填充指定颜色（B,G,R)
top_size, bottom_size, left_size, right_size = 50, 50, 50, 50
replicate = cv2.copyMakeBorder(img, top_size, bottom_size, left_size, right_size, cv2.BORDER_CONSTANT, value=[255,255,0])

#镜子填充
reflect = cv2.copyMakeBorder(img, top_size, bottom_size, left_size, right_size, cv2.BORDER_REFLECT)

cv2.imshow('常规填充', replicate)
cv2.imshow('镜像填充', reflect)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import matplotlib.pyplot as plt  #导入画图库
img = cv2.imread('graduate.png')

#常熟填充：边缘填充指定颜色（B,G,R)
top_size, bottom_size, left_size, right_size = 50, 50, 50, 50
replicate = cv2.copyMakeBorder(img, top_size, bottom_size, left_size, right_size, cv2.BORDER_CONSTANT, value=[255,255,0])

#镜子填充
reflect = cv2.copyMakeBorder(img, top_size, bottom_size, left_size, right_size, cv2.BORDER_REFLECT)

#把BGR转换为RGB格式
replicate_rgb = cv2.cvtColor(replicate, cv2.COLOR_BGR2RGB)
reflect_rgb = cv2.cvtColor(reflect, cv2.COLOR_BGR2RGB)

#用matplotlib窗口显示
plt.figure()
plt.title('常规填充（Constant）')
plt.imshow(replicate_rgb)

plt.figure()
plt.title('镜像显示（Reflect）')
plt.imshow(reflect_rgb)

plt.show()

# cv2.imshow('常规填充', replicate)
# cv2.imshow('镜像填充', reflect)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

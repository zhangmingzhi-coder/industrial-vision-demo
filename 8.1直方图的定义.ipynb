```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('atmqq-drszl.jpg', 0)

#统计图像中各个像素值的分布情况
# bins = 256,表示分为256个区间，range=[0,256]表示灰度图的范围
hist = cv2.calcHist([img], [0], None, [256], [0, 256])

plt.figure()
plt.title('Grayscale Histogram')
plt.xlabel("Gray Vale")
plt.ylabel('Pixel Count')
plt.plot(hist)
plt.xlim([0, 256])
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('atmqq-drszl.jpg')
colors = ('b', 'g', 'r')
for i, col in enumerate(colors):
    hist = cv2.calcHist([img], [i], None, [256], [0, 256])
    plt.plot(hist, color=col)
plt.show()
```

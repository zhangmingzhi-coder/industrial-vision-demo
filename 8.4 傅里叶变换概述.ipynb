```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

img = cv2.imread('atmqq-drszl.jpg', 0)
imgs = img
#1.二维傅里叶变换
f = np.fft.fft2(img)

#2.频普平移（将低频变换到中心）
fshift = np.fft.fftshift(f)

#3.转换为振幅普（取对数以便显示）
magnitude_spectrum = 20 * np.log(np.abs(fshift))

fig, axes = plt.subplots(1, 2, figsize=(15,10))
axes[0].imshow(imgs)
axes[0].set_title('Source Image', fontsize=14)
axes[0].axis('off')

axes[1].imshow(img)
axes[1].set_title('Template', fontsize=14)
axes[1].axis('off')

plt.tight_layout()
plt.show()

```
